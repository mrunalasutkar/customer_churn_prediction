```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
data=pd.read_csv(r"D:\customer_churn_project\data\churn.csv")
```

```python
data=data.drop(columns="customerID")
```

```python
data.head(3)
```

```python
data["TotalCharges"]=pd.to_numeric(data["TotalCharges"],errors="coerce")
```

```python
data["TotalCharges"] = data["TotalCharges"].fillna(0)
```

```python
data["Churn"] = data["Churn"].map({"No": 0, "Yes": 1})
```

```python
binary_cols = ['gender', 'Partner', 'Dependents', 'PhoneService', 'MultipleLines',
               'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport',
               'StreamingTV', 'StreamingMovies', 'PaperlessBilling']
```

```python
for col in binary_cols:
    if col == 'gender':
        data[col] = data[col].map({"Female": 0, "Male": 1})
    elif col == 'MultipleLines':
        data[col] = data[col].map({"No": 0, "No phone service": 0, "Yes": 1})
    else:
        data[col] = data[col].map({"No": 0, "No internet service": 0, "Yes": 1})
```

```python
print(data['MultipleLines'].value_counts())
```

```python
data.to_csv(r"D:\customer_churn_project\data\churn_cleaned.csv", index=False)
print("✅ Cleaned data saved with correct preprocessing order")
```
